# VI. Множини (Set).

17. Створити множину без елементів. Після цого за допомогою методу додайте кілька різних елементів до множини. Чи множини є змінним типом даних?

```python
my_set = set()

my_set.add("Python")
my_set.add(42)

my_set.update([10.5, "Vibe", 42])

print("Вміст множини:", my_set)
```

Так, множини (set) є змінним (mutable) типом даних.

18. Створити множину, маючи список my_list = [1, 1, 2, 67, 67, 8, 9]. Пояснити, чому "зникли" деякі елементи.

```python
my_list = [1, 1, 2, 67, 67, 8, 9]

my_set = set(my_list)

print("Початковий список:", my_list)
print("Отримана множина:", my_set)
```

Елементи зникли тому, що множина зберігає лише унікальні значення, автоматично відкидаючи будь-які дублікати

19. Створіть дві множини. Продемонстуйте над ними операції: обʼєднання, різниці, пересічі та симетричної різниці. Використовуйте методи, що не змінюють множини, а створюють нові.

```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

union_set = set_a.union(set_b)

intersection_set = set_a.intersection(set_b)

difference_set = set_a.difference(set_b)

sym_diff_set = set_a.symmetric_difference(set_b)

print(f"Множина A: {set_a}")
print(f"Множина B: {set_b}")
print(f"Об'єднання: {union_set}")
print(f"Перетин: {intersection_set}")
print(f"Різниця (A - B): {difference_set}")
print(f"Симетрична різниця: {sym_diff_set}")
```

# VII. Словники (Dictionary).

20. Створіть пустий словник. До нього додайте чотири пари елементів такі, щоб їхні ключі були різних типів. Чи може список бути ключем? Чому?

```python
my_dict = {}

my_dict[1] = "Integer key"
my_dict["name"] = "String key"
my_dict[3.14] = "Float key"
my_dict[(1, 2)] = "Tuple key"

for key, value in my_dict.items():
    print(f"Тип ключа {type(key).__name__:} Значення: {value}")
```

Ні, список не може бути ключем словника. Ключ словника має бути незмінним, щоб його хеш залишався сталим, а список — це змінний тип, який може зламати механізм пошуку в пам'яті.
