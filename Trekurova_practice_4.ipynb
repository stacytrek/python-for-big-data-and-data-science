6. (7б) Визначте функцію, яка приймає iterable (ітерабельний обʼєкт, напр. список), що складається з 7ми рядкових значень - назв погодних умов (напр. "sunny") за кожен день у тижні. Повертає функція словник, де ключами є рядки з погодними умовами, а значеннями - кількість разів, скільки дана погода була протягом тижня (напр. {'sunny': 1, ...}). Викличте цю функцію та виведіть результат її роботи на екран.

```python
def calculate_weather(data_list):
    weather_counts = {}

    for condition in data_list:
        weather_counts[condition] = weather_counts.get(condition, 0) + 1

    return weather_counts

my_week_weather = ["sunny", "cloudy", "rainy", "sunny", "stormy", "windy", "sunny"]

result = calculate_weather(my_week_weather)
print(result)
```

# III. Decorators.

11. (11б) Визначте декоратор, який би виводив у консоль повідомлення про початок роботи і назву функції (напр. my_function._ _ name _ _), а після відпрацювання функції - повідомлення про закінчення роботи і назву функції, що задекорована.

```python
def log_execution(func):
    def wrapper(*args, **kwargs):
        print(f"Початок роботи функції: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Закінчення роботи функції: {func.__name__}")
        return result
    return wrapper

@log_execution
def say_hello(name, greeting="Привіт"):
    print(f"{greeting}, {name}!")

say_hello("Насте")
```

12. (4б) Задекоруйте двома різними способами функції з завдання 4 та 5 декоратором із завдання 11. (**УВАГА**: не декоруйте функції двічі!).

```python
#Спосіб 1
@log_execution
def divide_num(num1, num2):
    if num2 == 0:
        return None
    return num1 / num2

#Спосіб 2
def diff_num(num1, num2=1):
    return num1 - num2
diff_num = log_execution(diff_num)

#Тест
print(" Тест Способу 1 (Ділення) ")
res1 = divide_num(10, 2)
print(f"Результат ділення: {res1}\n")

print(" Тест Способу 2 (Віднімання) ")
res2 = diff_num(10)
print(f"Результат віднімання: {res2}")
```

15. (15б) Визначте функцію, яка рекурсивно реалізовує пошук числа у послідовності Фібоначчі за її індексом (нумерація починається з нуля). Наприклад, потрібно знайти перше число у послідовності, тоді функція прийматиме 0 у якості аргумента і повертатиме 1, якщо ж знаходити третє число (індекс 2) послідовності, то функція повертатиме 2, якщо ж знаходити десяте число (індекс 9), то функція повертатиме 55.

```python
def fibonacci(index):

    if index == 0 or index == 1:
        return 1
    return fibonacci(index - 1) + fibonacci(index - 2)

print(f"Індекс 0: {fibonacci(0)}")
print(f"Індекс 2: {fibonacci(2)}")
print(f"Індекс 9: {fibonacci(9)}")
```
