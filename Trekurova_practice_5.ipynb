5. (9б). Створіть клас 3D-фігура без атрибутів, з методом get_volume для отримання об'єму фігури, що повертає 0 та __add __, який приймає self та other в якості аргументів, а повертає суму об'ємів 3d-фігур self та other.

```python
class Figure3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        return self.get_volume() + other.get_volume()

fig1 = Figure3D()
fig2 = Figure3D()

print(f"Об'єм однієї фігури: {fig1.get_volume()}")
print(f"Сума об'ємів двох фігур: {fig1 + fig2}")
```

6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
import math

class Figure3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        return self.get_volume() + other.get_volume()

class Cube(Figure3D):
    def __init__(self, side):
        self.side = side

    def get_volume(self):
        return self.side ** 3

class Cylinder(Figure3D):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        return math.pi * (self.radius ** 2) * self.height

my_cube = Cube(3)
my_cylinder = Cylinder(2, 5)

total_volume = my_cube + my_cylinder

print(f"Об'єм куба: {my_cube.get_volume():.2f}")
print(f"Об'єм циліндра: {my_cylinder.get_volume():.2f}")
print(f"Сумарний об'єм: {total_volume:.2f}")
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
    def __init__(self, wallet_id, initial_balance=0.0):
        self.__id = wallet_id
        self.__balance = float(initial_balance)

    # Геттер для балансу
    @property
    def balance(self):
        return self.__balance

    # Геттер для ID
    @property
    def wallet_id(self):
        return self.__id

    # Метод для поповнення
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Поповнено на {amount}. Поточний баланс: {self.__balance}")
        else:
            print("Сума поповнення має бути додатною!")

    # Метод для зняття
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Знято {amount}. Залишок: {self.__balance}")
        elif amount > self.__balance:
            print("Недостатньо коштів на рахунку!")
        else:
            print("Сума зняття має бути додатною!")

my_wallet = EWallet("UA-777", 100.0)
print(f"Гаманець: {my_wallet.wallet_id}, Баланс: {my_wallet.balance}")

my_wallet.deposit(50)
my_wallet.withdraw(30)
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name}: {self.price} грн"

class ShoppingCart:
    def __init__(self):
        self.__items = []

#Додавання айтема
    def add_item(self, item):
        if isinstance(item, Item):
            self.__items.append(item)
            print(f"Додано: {item.name}")
        else:
            print("Помилка: Можна додавати лише об'єкти типу Item")

#Видалення айтема
    def remove_item(self, item_name):
        for item in self.__items:
            if item.name == item_name:
                self.__items.remove(item)
                print(f"Видалено: {item_name}")
                return
        print(f"Товар '{item_name}' не знайдено в кошику.")

    def show_items(self):
        if not self.__items:
            print("Кошик порожній.")
            return

        print("\nВміст кошика ")
        total = 0
        for item in self.__items:
            print(f"- {item}")
            total += item.price
        print(f"Загальна вартість: {total} грн\n")


cart = ShoppingCart()

# Створюємо товари
milk = Item("Молоко", 45.0)
bread = Item("Хліб", 22.5)

# Використовуємо методи
cart.add_item(milk)
cart.add_item(bread)
cart.show_items()

cart.remove_item("Молоко")
cart.show_items()
```

10. (13б). Створіть клас Vehicle з атрибутами brand (manufacturer) та production_age. Створіть ще один клас Engine з такими атрибутами, як type (ex. petrol) та size (volume, litres). Створіть клас Car, який успадковує обидва класи Vehicle та Engine. Продемонструйте використання множинної спадковості, створивши обʼєкт класу Car та отримавши mro для цього класу.

```python
class Vehicle:
    def __init__(self, brand, production_age):
        self.brand = brand
        self.production_age = production_age

class Engine:
    def __init__(self, engine_type, size):
        self.engine_type = engine_type
        self.size = size

class Car(Vehicle, Engine):
    def __init__(self, brand, production_age, engine_type, size, model):
        Vehicle.__init__(self, brand, production_age)
        Engine.__init__(self, engine_type, size)
        self.model = model

    def get_info(self):
        return (f"{self.brand} {self.model} ({self.production_age} р.), "
                f"двигун: {self.engine_type} {self.size}L")

my_car = Car("Toyota", 2023, "Petrol", 2.5, "Camry")

print(f"Інформація про авто: {my_car.get_info()}")

print("\nПорядок вирішення методів (MRO):")
for cls in Car.mro():
    print(cls)
```

Ітерабельний об'єкт — це «колекція-сховище» (як книга чи список), яку можна перебрати, а ітератор — це «активний покажчик» (як закладка), який пам'ятає поточну позицію і видає елементи по одному за допомогою команди next().

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class WeekdayIterator:
    def __init__(self):
        self.days = [
            "Понеділок", "Вівторок", "Середа",
            "Четвер", "П'ятниця", "Субота", "Неділя"
        ]
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.days):
            day = self.days[self.index]
            self.index += 1
            return day
        else:
            raise StopIteration

week = WeekdayIterator()

for day in week:
    print(day)
```
